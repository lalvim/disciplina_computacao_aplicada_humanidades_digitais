## 2. Datas, cobertura e três perguntas distintas

Antes do cálculo, registre o que a data representa: ocorrência, publicação, registro ou
digitalização; e qual sua precisão: dia, mês, ano ou intervalo. Uma série anual mistura
processos diferentes quando essas definições mudam silenciosamente.

- **contagem:** quantos eventos foram registrados?;
- **taxa:** quantos eventos por população exposta?;
- **crescimento:** quanto o valor mudou em relação ao período anterior?

Uma taxa melhora a comparabilidade do denominador, mas não corrige a mudança de
cobertura documentada em 1990.


A taxa é $r_t=(y_t/N_t)\times100000$. O crescimento anual é
$g_t=((y_t-y_{t-1})/y_{t-1})\times100$. Crescimento instável pode resultar de base
pequena; compare também níveis absolutos.
