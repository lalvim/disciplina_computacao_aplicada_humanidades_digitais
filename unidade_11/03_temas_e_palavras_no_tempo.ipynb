## 1. Normalizar pela oportunidade de ocorrência

Antes de comparar palavras entre anos, dividimos as menções pelo número de documentos. A curva resultante descreve frequência relativa; a interpretação histórica virá depois, com metadados e leitura de concordâncias.


```python
import pandas as pd,matplotlib.pyplot as plt
d=pd.read_csv("dados/termos_por_ano.csv")
for termo in ["patrimonio","digital"]:
 d[f"taxa_{termo}"]=100*d[f"mencoes_{termo}"]/d.documentos
 d[f"media_{termo}"]=d[f"taxa_{termo}"].rolling(5,center=True).mean()
display(d.loc[d.ano.isin([1960,1990,2000,2020])].round(2))
d.plot(x="ano",y=["media_patrimonio","media_digital"],figsize=(10,4));plt.ylabel("Menções por 100 documentos — média móvel de 5 anos");plt.tight_layout();plt.show()
```
