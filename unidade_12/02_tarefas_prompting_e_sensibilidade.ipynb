## 2. Protocolo de prompt e registro do experimento

Uma instrução auditável deve declarar: tarefa; corpus permitido; unidade de saída;
formato; obrigação de citar evidência; comportamento quando a informação estiver
ausente; e critérios de avaliação.

| Componente | Exemplo controlado |
|---|---|
| tarefa | extrair o número de matrículas |
| corpus | somente F001–F010 |
| formato | `id_documento; valor; trecho` |
| ausência | escrever `não consta` |
| evidência | copiar o trecho e citar o identificador |

Congele variantes antes de ver os resultados e registre texto integral do prompt,
modelo/versão quando houver, configuração, data, entrada e saída. Assim, sensibilidade
não vira seleção retrospectiva da resposta preferida.
