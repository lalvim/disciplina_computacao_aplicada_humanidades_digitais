## Sequência lógica deste notebook

Como nos slides, o raciocínio parte de um exemplo ou problema concreto e só
então introduz a técnica. Leia e execute as etapas nesta ordem:

1. **Alegação**
2. **Conceito e medida**
3. **Inferência**
4. **Contexto de alcance**
5. **Ameaça**
6. **Evidência necessária**

A saída de cada etapa fornece a justificativa para a seguinte; o resultado
computacional só se torna evidência depois de interpretado e confrontado com
casos, fontes ou limites.


## 1. Da alegação aos três tipos de validade

Começamos por uma conclusão concreta e percorremos conceito, medida, inferência e alcance. A matriz seguinte não é uma lista abstrata: cada coluna testa um elo diferente da mesma alegação.


<!-- complemento-sequencia -->
## 2. Uma ameaça exige evidência e ação

Validade de construto pergunta se a medida representa o conceito; validade interna,
se explicações alternativas permanecem; validade externa, até onde a conclusão viaja.
Para cada ameaça, registre evidência que a testaria e como o resultado alteraria a
conclusão. Uma lista sem consequência não constitui auditoria.


## Matriz de auditoria preenchível

| Alegação | Construto | Interna | Externa | Evidência necessária |
|---|---|---|---|---|
| “Tema aumentou” | palavra mede tema? | cobertura mudou? | vale fora do acervo? | concordâncias, metadados e outro corpus |
| “Grupo difere” | categoria é coerente? | composição explica? | outros grupos/períodos? | casos, controles e replicação |
