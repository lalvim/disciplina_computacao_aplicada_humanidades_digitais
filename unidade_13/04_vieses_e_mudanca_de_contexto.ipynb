## 1. Localizar diferenças por grupo e período

Calculamos desempenho estratificado para localizar onde a média falha. A lacuna é um sinal para retornar à composição e aos casos; não identifica sozinha o mecanismo nem prova discriminação algorítmica.


```python
import pandas as pd
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
d=pd.read_csv("dados/predicoes_auditoria.csv")
def medir(g):return pd.Series({"n":len(g),"acurácia":accuracy_score(g.classe_ouro,g.classe_prevista),"F1":f1_score(g.classe_ouro,g.classe_prevista)})
por_periodo=d.groupby("periodo").apply(medir,include_groups=False);por_origem=d.groupby("origem").apply(medir,include_groups=False)
display(por_periodo.round(3));display(por_origem.round(3))
```

## 2. Da lacuna ao mecanismo, teste e mitigação

O viés pode entrar na seleção dos casos, na formação do corpus, na anotação, na
representação, no algoritmo ou no uso. Para não saltar da diferença à causa, associe
cada hipótese a uma verificação e a uma ação coerente.

| Mecanismo rival | Teste possível | Mitigação coerente |
|---|---|---|
| OCR pior no período antigo | comparar erros por qualidade de OCR | corrigir/amostrar OCR e reavaliar |
| gêneros diferentes por origem | estratificar por gênero documental | reponderar ou limitar o alcance |
| guia ambíguo em um grupo | dupla anotação e desacordos por grupo | revisar guia e readjudicar |
| mudança histórica de vocabulário | ler falsos negativos em períodos distintos | adaptar representação e testar transporte |

Mitigação deve responder ao mecanismo confirmado; igualar uma métrica sem compreender
a origem da lacuna pode esconder o problema.


## Mudança entre períodos e contextos

Queda no período antigo pode refletir grafia, OCR, gêneros ou anotação. Queda no
Interior pode refletir menor cobertura ou diferença substantiva. Audite composição,
qualidade e erros antes de atribuir o problema ao grupo.
