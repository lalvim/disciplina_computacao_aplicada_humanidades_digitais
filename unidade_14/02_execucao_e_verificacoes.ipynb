```python
# @title Preparação do ambiente — execute esta célula no Google Colab
from pathlib import Path
import importlib.util
import os
import subprocess
import sys

URL_REPOSITORIO = 'https://github.com/lalvim/disciplina_computacao_aplicada_humanidades_digitais.git'
REPOSITORIO = Path(
    "/content/disciplina_computacao_aplicada_humanidades_digitais"
)
PASTA_UNIDADE = REPOSITORIO / 'unidade_14'

try:
    import google.colab  # type: ignore  # noqa: F401
    EM_COLAB = True
except ImportError:
    EM_COLAB = False

if EM_COLAB:
    if not (REPOSITORIO / ".git").exists():
        subprocess.run(
            [
                "git",
                "clone",
                "--depth",
                "1",
                "--branch",
                "main",
                URL_REPOSITORIO,
                str(REPOSITORIO),
            ],
            check=True,
        )

    PACOTES_COLAB = []
    ausentes = [
        especificacao
        for modulo, especificacao in PACOTES_COLAB
        if importlib.util.find_spec(modulo) is None
    ]
    if ausentes:
        subprocess.run(
            [sys.executable, "-m", "pip", "install", "-q", *ausentes],
            check=True,
        )

    os.chdir(PASTA_UNIDADE)
    print("Ambiente preparado em:", Path.cwd())
else:
    print("Ambiente local: nenhuma clonagem necessária.")
```

```python
import hashlib,json,platform,pandas as pd,numpy as np
from pathlib import Path
arquivo=Path("dados/resultados_comunicacao.csv");d=pd.read_csv(arquivo)
hash_sha256=hashlib.sha256(arquivo.read_bytes()).hexdigest()
ambiente={"python":platform.python_version(),"pandas":pd.__version__,"numpy":np.__version__}
checks={"linhas":len(d)==3,"colunas":list(d.columns)==["analise","estimativa","limite_inferior","limite_superior","alcance"],"intervalos_validos":((d.limite_inferior<=d.estimativa)&(d.estimativa<=d.limite_superior)).all()}
print("SHA-256:",hash_sha256);print("Ambiente:",ambiente);print("Checks:",checks);assert all(checks.values())
```

Hash detecta mudança de bytes, não garante qualidade ou autenticidade. Versões exatas
favorecem repetição; faixas controladas favorecem manutenção. Registre sistema e data,
mas não publique caminhos pessoais, tokens ou segredos.
